# Compare activations in biased and unbiased prompt sets

This notebook runs the same analysis on both available activation caches and saves results to separate `biased/` and `unbiased/` folders. The biased set uses the open prompt `a picture of a {race} {gender}`. The unbiased set uses the prompt-engineered neutral-expression, no-facial-covering, white-T-shirt, white-background template.

For a direct comparison, both use the mean across images of each image's spatial-max activation. The neutral cache only stores spatial maxima, so the biased cache's spatial-max matrix is used too. There are 50 neutral and 100 open-prompt examples per race/gender group. The neutral cache was generated independently from the saved baseline PNGs and has no per-row seed metadata.

Each dataset gets 12 figures: one per race/gender group, with a 5×2 grid of that group's top ten features and 12 category bars per feature. Rankings are dataset-specific. These are descriptive prompt-group associations, not semantic or causal results.

```python
from pathlib import Path
from runpy import run_path

NOTEBOOK_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
                     if p.name == 'sae-demo-notebooks' and (p / 'sae-activation').exists())
SCRIPT = NOTEBOOK_ROOT / 'sae-activation' / 'compare_mean_activations_biased_unbiased.py'
run_path(str(SCRIPT), run_name='__main__')
```

## Results

The script writes 12 figures to `mean_activations_by_race_gender_comparison/biased/` and 12 to `mean_activations_by_race_gender_comparison/unbiased/`. The combined activation values are in `mean_activations_by_race_gender_comparison/mean_activations_biased_unbiased.csv`.
