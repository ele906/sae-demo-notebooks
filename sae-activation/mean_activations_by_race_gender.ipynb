# Mean SAE activations by race and gender

This notebook creates 12 separate figures, one for each race/gender prompt group. Each figure uses a 5×2 grid for that group's top ten features; each feature panel has 12 bars comparing mean activation across all race/gender groups. Men use a darker shade and women a lighter shade of the same race color.

It uses the 1,200-image open-prompt dataset (`a picture of a {race} {gender}`), not the engineered neutral-shirt/background dataset. Rankings and means use all 1,200 images, so this is an exploratory descriptive comparison. Feature IDs refer to the existing pretrained SAE; activation differences do not establish feature meaning or causation.

```python
from pathlib import Path
from runpy import run_path
from IPython.display import display, Image

NOTEBOOK_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
                     if p.name == 'sae-demo-notebooks' and (p / 'sae-activation').exists())
SCRIPT = NOTEBOOK_ROOT / 'sae-activation' / 'plot_mean_activations_by_group.py'
run_path(str(SCRIPT), run_name='__main__')
```

```python
FIGURES = NOTEBOOK_ROOT / 'sae-activation' / 'mean_activations_by_race_gender'
for path in sorted(FIGURES.glob('*.png')):
    print(path.stem.replace('_', ' ').title())
    display(Image(filename=str(path)))
```
